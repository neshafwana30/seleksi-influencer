```python
from pathlib import Path

def find_repo_root():
    current = Path.cwd()
    for _ in range(6):
        if (current / '01_pengumpulan_data').is_dir() and (current / '03_Indobert').is_dir():
            return current
        if current.parent == current:
            break
        current = current.parent
    return None

REPO_ROOT = find_repo_root()

CHECKPOINTS = {
    '3class': REPO_ROOT / '03_Indobert' / 'models' / '3class' / 'final' / 'checkpoints' / 'checkpoint-692',
    'binary': REPO_ROOT / '03_Indobert' / 'models' / 'binary' / 'final' / 'checkpoints' / 'checkpoint-258',
}

WAJIB = ['config.json', 'model.safetensors']
OPSIONAL = ['pytorch_model.bin', 'trainer_state.json', 'training_args.bin',
            'tokenizer_config.json', 'vocab.txt', 'special_tokens_map.json']

for scheme, path in CHECKPOINTS.items():
    print(f'\n=== {scheme} ===')
    print(f'Path: {path}')
    if not path.exists():
        print('  FOLDER GA KETEMU!')
        continue

    isi = sorted(p.name for p in path.iterdir())
    print(f'  Isi folder: {isi}')

    print('  Cek file wajib:')
    semua_ada = True
    for f in WAJIB:
        ada = (path / f).exists()
        ukuran = f'({(path / f).stat().st_size / 1e6:.1f} MB)' if ada else ''
        print(f'    [{"OK" if ada else "HILANG"}] {f} {ukuran}')
        if not ada:
            semua_ada = False

    print('  Cek file opsional (tokenizer dkk, kalau ada lumayan ga perlu download ulang):')
    for f in OPSIONAL:
        ada = (path / f).exists()
        print(f'    [{"ADA" if ada else "-"}] {f}')

    if semua_ada:
        import json
        with open(path / 'config.json', encoding='utf-8') as fh:
            cfg = json.load(fh)
        print(f'  num_labels di config.json: {cfg.get("num_labels", "?")} '
              f'(harusnya {"2 buat binary" if scheme == "binary" else "3 buat 3class"})')
        print(f'  model_type: {cfg.get("model_type", "?")}')
        print('  -> Checkpoint ini KELIHATANNYA LENGKAP, bisa dipake buat inference.')
    else:
        print('  -> Checkpoint ini KURANG LENGKAP, ga bisa langsung dipake.')
```
